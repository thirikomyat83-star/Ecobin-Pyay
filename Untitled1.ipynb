```python
from roboflow import Roboflow
import uuid
import os

rf = Roboflow(api_key="c8m0rs2c1okv1yH3LrL6")
vision_model = rf.workspace("grad-jhbv7").project("waste-classification-irwkg").version(1).model

@app.post("/api/scan")
async def scan_waste(file: UploadFile = File(...)):
    # Frontend မှ ပို့လာသော ပုံကို ယာယီသိမ်းဆည်းခြင်း
    temp_filename = f"temp_{uuid.uuid4()}.jpg"
    with open(temp_filename, "wb") as buffer:
        buffer.write(await file.read())

    try:
        # Roboflow Model သို့ ပုံပို့၍ စစ်ဆေးခြင်း
        prediction = vision_model.predict(temp_filename, confidence=40).json()
        
        # ယာယီပုံကို ပြန်ဖျက်ခြင်း
        os.remove(temp_filename)

        if not prediction['predictions']:
            return {"status": "failed", "message": "အမှိုက်အမျိုးအစားကို သေချာစွာ မဖတ်နိုင်ပါ။"}

        # အသေချာဆုံး (Confidence အများဆုံး) အဖြေကို ယူခြင်း
        best_match = prediction['predictions'][0]
        class_name = best_match['class'].lower()
        
        # Point တွက်ချက်ခြင်း
        detected_label = "Unknown"
        points, co2_saved = 0, 0.0

        if "plastic" in class_name or "bottle" in class_name:
            detected_label, points, co2_saved = "ပလတ်စတစ်ဘူး", 50, 0.08
        elif "can" in class_name or "metal" in class_name:
            detected_label, points, co2_saved = "အချိုရည် သံဘူး", 80, 0.15
        elif "glass" in class_name:
            detected_label, points, co2_saved = "ဖန်ပုလင်း", 100, 0.3
        else:
            detected_label, points, co2_saved = class_name, 10, 0.05

        return {
            "status": "success",
            "name": detected_label,
            "pts": points,
            "co2": co2_saved,
            "label": f"{detected_label} (AI Verified)"
        }
    except Exception as e:
        print("Scan Error:", e)
        if os.path.exists(temp_filename): os.remove(temp_filename)
        raise HTTPException(status_code=500, detail="Scan ဖတ်ရာတွင် အခက်အခဲရှိနေပါသည်။")
```
